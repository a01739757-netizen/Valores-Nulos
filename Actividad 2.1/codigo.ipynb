```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    '/Users/jos/Downloads/Plataformas de analítica de negocios para organizaciones/VALORES_NULOS/inmuebles2.csv',
    header=1
)
```

```python
#ver valores nulos
resumen = pd.DataFrame({
    'nulos':      df.isnull().sum(),
})

resumen = resumen[resumen['nulos'] > 0]  
resumen  
```

```python
#Eliminar columnas innecesarias
#axis = 1 para columnas
#axis = 0 para filas
df = df.drop(columns=['0'], errors='ignore')
```

```python
# columnas categoricas con pocos se vana moda
df['Fecha Alta']  = df['Fecha Alta'].fillna(df['Fecha Alta'].mode()[0])
df['Tipo']        = df['Tipo'].fillna(df['Tipo'].mode()[0])
df['Operación']   = df['Operación'].fillna(df['Operación'].mode()[0])
df['Provincia']   = df['Provincia'].fillna(df['Provincia'].mode()[0])
```

```python
# columnas numericas con pocos nulos van a mediana
df['Superficie']   = df['Superficie'].fillna(df['Superficie'].median())
df['Precio Venta'] = df['Precio Venta'].fillna(df['Precio Venta'].median())
```

```python
# columnas con nulos sin info 
df['Fecha Venta'] = df['Fecha Venta'].fillna('Sin fecha')
df['Vendedor']    = df['Vendedor'].fillna('Sin asignar')
```

```python

print("Nulos restantes:")
print(df.isnull().sum())
print("Registros totales:", {len(df)})
```

```python
#  archivo limpio
df.to_csv('archivo.csv', index=False)
```
