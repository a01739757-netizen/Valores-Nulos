```python
import pandas as pd
import numpy as np

# cargamos el excel tal cual viene, sin tocar nada, para ver su estructura real
df = pd.read_excel("Análisis_de_Piso.xlsx", sheet_name="Hoja1", header=None)

print("Filas x Columnas:", df.shape)
df.head(10)
```

```python
# valores nulos por columna (columnas numeradas 0,1,2... porque aun no le ponemos nombres)
df.isnull().sum()
```

```python
# las primeras 3 filas eran encabezado combinado -> se saltan
# (no se renombran las columnas, no hace falta: se trabaja por posicion)
df = pd.read_excel("Análisis_de_Piso.xlsx", sheet_name="Hoja1", header=None, skiprows=3)

# los "vacios" del excel vienen como \xa0, "-" o " ", se pasan a NaN real
df = df.replace(["\xa0", "-", " ", ""], np.nan)
df.head()
```

```python
# columna 0 = Asesor: si no tiene nombre real (total/sistema, ej. "sys_competition") -> se elimina
df[0] = df[0].astype(str).str.strip()
df = df[~df[0].str.lower().isin(["nan", "", "sys_competition"])]

# el resto de columnas son numericas
cols = df.columns.drop(0)
df[cols] = df[cols].apply(pd.to_numeric, errors="coerce")

# columnas de Raking (18, 19, 41, 49, 50): sin raking -> ultimo lugar (max + 1)
raking = [18, 19, 41, 49, 50]
df[raking] = df[raking].fillna(df[raking].max() + 1)

# el resto: sin actividad ese mes -> 0
otras = cols.drop(raking)
df[otras] = df[otras].fillna(0)

print("Nulos restantes:", df.isnull().sum().sum())
```

```python
# guardamos el archivo 1 ya limpio
df.to_excel("Analisis_de_Piso_limpio.xlsx", index=False)
```

```python
# valores nulos por columna (columnas numeradas 0,1,2... porque aun no le ponemos nombres)
df.isnull().sum()
```

```python
# las primeras 3 filas eran encabezado combinado -> se saltan
# (no se renombran las columnas, no hace falta: se trabaja por posicion)
df = pd.read_excel("Ventas_por_Asesor.xlsx", sheet_name="Hoja1", header=None, skiprows=3)

# los "vacios" del excel vienen como \xa0, "-" o " ", se pasan a NaN real
df = df.replace(["\xa0", "-", " ", ""], np.nan)
df.head()
```

```python
# columna 0 = Asesor: si no tiene nombre real, o es la fila "Total general" (es un agregado, no un asesor) -> se elimina
df[0] = df[0].astype(str).str.strip()
df = df[~df[0].str.lower().isin(["nan", "", "total general", "total"])]

# el resto de columnas son numericas
cols = df.columns.drop(0)
df[cols] = df[cols].apply(pd.to_numeric, errors="coerce")

# columnas de Raking (32, 34, 36, 38): sin raking -> ultimo lugar (max + 1)
raking = [32, 34, 36, 38]
df[raking] = df[raking].fillna(df[raking].max() + 1)

# el resto: sin ventas ese mes/periodo -> 0
otras = cols.drop(raking)
df[otras] = df[otras].fillna(0)

print("Nulos restantes:", df.isnull().sum().sum())
```

```python
# guardamos el archivo 2 ya limpio
df.to_excel("Ventas_por_Asesor_limpio.xlsx", index=False)
```
